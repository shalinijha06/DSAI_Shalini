```python
import spacy 
nlp=spacy.load("en_core_web_sm")
```

```python
#tokenization
text = "artificoal intelligence is amazing."
doc=nlp(text)
```

```python
for token in doc:
    print(token.text, token.pos)
```

```python
#lemmatization
for token in doc:
    print(token.text, token.lemma)
```

```python
#Named Entity Recognition(NER)
text = "Elon Musk is a businessman and entrepreneur."
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
#Tokenization
from textblob import TextBlob

blob = TextBlob("Python is easy  to learn.")

print(blob.words)
print(blob.sentences)
```

```python
#spell correction
blob = TextBlob("i havv a dreem")
print(blob.correct())
```

```python
#pos tagging
blob = TextBlob("ram teaches python")
print(blob.tags)
```

```python
#noun phrase
blob = TextBlob("machine learning and artificial intelligence are popular fields")
print(blob.noun_phrases)
```
