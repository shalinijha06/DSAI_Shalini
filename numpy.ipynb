```python
import numpy as np

# Student Marks Dataset (out of 100)
marks = np.array([45, 67, 89, 90, 56, 78, 88, 92, 61, 74,
                  83, 95, 52, 68, 71, 85, 77, 64, 58, 81])

print("Dataset:")
print(marks)
```

```python
print(np.mean(marks))
```

```python
print(np.median(marks))
```

```python
# 3. Mode
values, counts = np.unique(marks, return_counts=True)
max_count = np.max(counts)

if max_count == 1:
    print("3. Mode: No mode (all values occur once)")
else:
    mode = values[counts == max_count]
    print("3. Mode:", mode)
```

```python
print(np.max(marks))
print(np.min(marks))
```

```python
print(np.max(marks) - np.min(marks))
```

```python
print(np.var(marks))
```

```python
print(np.std(marks))
```

```python
print(np.percentile(marks,25))
```

```python
print(np.percentile(marks,50))
```

```python
print(np.percentile(marks,75))
```

```python
Q1 = np.percentile(marks, 25)
Q3 = np.percentile(marks, 75)

print("IQR =", Q3 - Q1)
```

```python
mean = np.mean(marks)
print(np.sum(marks > mean))
```

```python
print(np.sum(marks < 60))
```

```python
total = np.sum(marks)
average = total / len(marks)

print("Sum =", total)
print("Average =", average)
```

```python
#ascending
print(np.sort(marks))
```
