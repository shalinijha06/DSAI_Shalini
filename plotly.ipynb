```python
import plotly.express as px
```

```python
fig = px.scatter(df,
                 x = "Sales",
                 y = "Profit",
                 title ="sales vs profit")

fig.show()      
```

```python
sales_data = df.groupby("Product")["Sales"].sum().reset_index()

fig = px.bar(sales_data,
    x="Product",
    y="Sales",
    title="Total Sales for Each Product",
    color="Product"
)

fig.show()
```

```python
fig = px.pie(
    sales_data,
    names="Product",
    values="Sales",
    title="Sales Distribution by Product"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Profit",
    y="Sales",
    size="Quantity",
    color="Product",
    title="Sales vs Profit"
)

fig.show()
```

```python
fig = px.histogram(df,
              x="Sales",
              nbins=10,
             )
fig.show()
```

```python
fig = px.box(
    df,
    x="Product",
    y="Sales",
    color="Product",
    title="Sales Distribution of Each Product"
)

fig.show()
```

```python
city_sales = df.groupby("City", as_index=False)["Sales"].sum()

fig = px.bar(
    city_sales,
    x="City",
    y="Sales",
    title="Total Sales of Each City",
    color="City"
)

fig.show()
```

```python
avg_profit = df.groupby("Product", as_index=False)["Profit"].mean()

fig = px.line(
    avg_profit,
    x="Product",
    y="Profit",
    title="Average Profit of Each Product",
    markers=True
)

fig.show()
```
