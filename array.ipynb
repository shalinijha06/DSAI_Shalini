```python
import numpy as np

# Matrix A
A = np.array([
    [2, 4, 6],
    [8, 10, 12],
    [14, 16, 18]
])

# Matrix B
B = np.array([
    [1, 3, 5],
    [7, 9, 11],
    [13, 15, 17]
])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
```

```python
print(A.shape)
print(B.shape)
print(A.size)
print(B.size)
print(A.ndim)
print(B.ndim)
```

```python
print(A + B)
```

```python
print(A - B)
```

```python
print(A * B)
```

```python
print(np.matmul(A, B))
```

```python
print(A.T)
print(B.T)
```

```python
print(np.linalg.det(A))
```

```python
print(np.linalg.inv(A))
```

```python
print("Trace of A:", np.trace(A))
print("Trace of B:", np.trace(B))
```

```python
print("Maximum:", np.max(A))
print("Minimum:", np.min(A))
print("Mean:", np.mean(A))
print("Sum:", np.sum(A))
```

```python
print(np.sum(B, axis=1))

print(np.sum(B, axis=0))
```

```python
print(A[0])
print(A[:,1])
print(A[-1])
print(A[-1,-1])
```

```python
print(A.reshape(1,9))
print(B.flatten())
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```

```python
print(np.hstack((A,B)))
print(np.vstack((A,B)))
```
